#Exercicio 1
Utilizando o built-in method input(), crie um programa que receba a altura e o peso de uma pessoa e imprima na tela o IMC da mesma.

```python
# Solicita ao usuário o peso em quilogramas (tipo float para permitir decimais)
peso = float(input("Digite seu peso (kg): "))

# Solicita ao usuário a altura em metros (tipo float para permitir decimais)
altura = float(input("Digite sua altura (m): "))

# Calcula o Índice de Massa Corporal (IMC) usando a fórmula: peso / (altura²)
imc = peso / (altura ** 2)   # ** é o operador de exponenciação em Python

# Exibe o resultado formatado com duas casas decimais
print(f"Seu IMC é: {imc:.2f}")  # f-string permite inserir variáveis diretamente na string
```

#Exercicio 3
Desenhe um código que extraia o domínio de um e-mail informado.

```python
# Solicita ao usuário que digite um endereço de e-mail
email = input("Digite seu e-mail: ")

# Divide o e-mail em duas partes usando o caractere "@" como separador.
# A parte antes do "@" é o índice 0 (nome do usuário) e a parte depois é o índice 1 (domínio).
# [1] seleciona o domínio (ex: "gmail.com", "outlook.com").
dominio = email.split("@")[1]

# Exibe o domínio extraído, formatado dentro da string
print(f"O domínio do e-mail é: {dominio}")
```

#Exercicio 4
Faça um programa para uma loja de tintas. A pessoa informa a área em m2 que deseja pintar, e o script calculará a quantidade de latas de tinta que a pessoa deve comprar e o valor. Considere que cada litro de tinta pinta 3m2, que cada lata contém 18L e que custa R$ 80.



```python

# Solicita ao usuário a área a ser pintada em metros quadrados (float permite decimais)
area_m2 = float(input("Informe a área a ser pintada (m²): "))

# Constantes definidas pelo problema:
# Quantos litros de tinta são necessários por metro quadrado
LITROS_POR_M2 = 3
# Capacidade de cada lata (em litros)
LATA_LITROS = 18
# Preço de cada lata
PRECO_LATA = 80

# Calcula a quantidade total de litros necessários com base na área
litros_necessarios = area_m2 / LITROS_POR_M2

# Calcula quantas latas inteiras seriam necessárias (divisão inteira)
# Ex: 20 litros / 18 = 1 (pois int() trunca)
latas_necessarias = int(litros_necessarios / LATA_LITROS)

# Verifica se a divisão é exata (resto zero)
if (litros_necessarios % LATA_LITROS == 0):
    # Se for exato, mantém o valor calculado (não precisa adicionar nada)
    latas_necessarias = latas_necessarias
else:
    # Se houver resto, significa que uma lata extra é necessária para cobrir o restante
    latas_necessarias = latas_necessarias + 1

# Calcula o custo total multiplicando número de latas pelo preço unitário
valor_total = latas_necessarias * PRECO_LATA

# Exibe os resultados formatados
print(f"Área: {area_m2} m²")
print(f"Litros necessários: {litros_necessarios:.2f} L")      # Duas casas decimais nos litros
print(f"Latas de 18L necessárias: {latas_necessarias}")       # Número inteiro de latas
print(f"Valor total: R$ {valor_total:.2f}")                  # Valor com duas casas decimais
```

#Exercicio 5
Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês. Calcule e mostre o total do seu salário no referido mês, sabendo-se que são descontados 11% para o Imposto de Renda, 8% para o INSS e 5% para o sindicato, faça um programa que nos dê:
1. Salário bruto.
2. Quanto pagou de Imposto de renda
3. Quanto pagou ao INSS.
4. Quanto pagou ao sindicato.
5. O salário líquido.

```python
# Solicita ao usuário o valor ganho por hora (permite decimais com float)
valor_hora = float(input("Quanto você ganha por hora? R$ "))

# Solicita o total de horas trabalhadas no mês
horas_trabalhadas = float(input("Número de horas trabalhadas no mês: "))

# Calcula o salário bruto (sem descontos)
salario_bruto = valor_hora * horas_trabalhadas

# Calcula o desconto do Imposto de Renda: 11% do salário bruto
ir = salario_bruto * 0.11

# Calcula o desconto do INSS: 8% do salário bruto
inss = salario_bruto * 0.08

# Calcula o desconto do sindicato: 5% do salário bruto
sindicato = salario_bruto * 0.05

# Calcula o salário líquido subtraindo todos os descontos do salário bruto
salario_liquido = salario_bruto - ir - inss - sindicato

# Exibe os resultados de forma organizada, com duas casas decimais
print(f"1. Salário bruto: R$ {salario_bruto:.2f}")
print(f"2. Imposto de Renda (11%): R$ {ir:.2f}")
print(f"3. INSS (8%): R$ {inss:.2f}")
print(f"4. Sindicato (5%): R$ {sindicato:.2f}")
print(f"5. Salário líquido: R$ {salario_liquido:.2f}")
```

#Exercicio 6
Faça um programa que diz se o ano é bissexto ou não.
Um ano é bissexto quando:
- De 4 em 4 anos é bissexto
- Exceto os múltiplos de 100
- A não ser que também sejam múltiplos de 400.


```python
# Solicita ao usuário um ano (como número inteiro) para verificar se é bissexto
ano = int(input("Digite um ano (ex: 2024): "))

# Regra do ano bissexto (calendário gregoriano):
# - É divisível por 4 E não é divisível por 100, OU
# - É divisível por 400
# Condição composta com operadores lógicos: and / or
if (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0):
    # Caso a condição seja verdadeira, o ano é bissexto
    print(f"O ano {ano} é BISSEXTO!")
else:
    # Caso contrário, não é bissexto
    print(f"O ano {ano} NÃO é bissexto.")
```

#Exercicio 7
Faça um programa que:
- Peça ao usuário dois valores
- Salve cada valor em uma variavel
- Troque os valores das variaveis
Ex:
A = 50
B =100
O novo valor de a é 100.
O novo valor de b é 50.

```python
# Solicita ao usuário o primeiro valor (como string, pode ser texto ou número)
valor1 = input("Digite o primeiro valor: ")

# Solicita ao usuário o segundo valor
valor2 = input("Digite o segundo valor: ")

# Atribui os valores às variáveis 'a' e 'b' para facilitar a troca
a = valor1
b = valor2

# Exibe os valores originais antes da troca
# O '\n' insere uma quebra de linha antes da mensagem
print(f"\nValores originais -> a = {a} | b = {b}")

# Início da lógica de troca usando variável auxiliar
# Guarda o valor de 'a' em uma variável temporária 'auxiliar'
auxiliar = a

# Copia o valor de 'b' para 'a' (agora 'a' recebe o valor original de 'b')
a = b

# Passo 3: Copia o valor guardado (original de 'a') para 'b'
b = auxiliar
# Fim da troca

# Exibe os novos valores após a troca
print(f"\nO novo valor de a é {a}.")
print(f"O novo valor de b é {b}.")
```

#Exercicio 8
Peça que o usuário digite um numero qualquer e diga se o número é impar ou par.

```python
# Solicita ao usuário um número inteiro e converte a entrada para o tipo int
numero = int(input("Digite um número inteiro: "))

# Verifica se o número é par:
# O operador % retorna o resto da divisão por 2.
# Se o resto for 0, o número é divisível por 2 → par.
if numero % 2 == 0:
    # Caso verdadeiro: exibe mensagem informando que o número é PAR
    print(f"O número {numero} é PAR.")
else:
    # Caso contrário (resto diferente de 0): o número é ÍMPAR
    print(f"O número {numero} é ÍMPAR.")
```

#Exercicio 9
Crie um programa que:
- Peça 3 valores ao usuário
- Ordene esses valores em ordem crescente.
- O programa deve ser executado usando lógica condicional (IF, ELIF e ELSE)

Exemplo
Digite 3 números:
num1 = 10
num2 = 5
num3 = 20
Os números digitados foram:
5
10
20

```python
# Solicita três números ao usuário (convertendo para float para aceitar decimais)
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))

# --- Estrutura de decisão para encontrar menor, meio e maior ---
# Caso 1: num1 é o menor (quando é <= num2 e <= num3)
if num1 <= num2 and num1 <= num3:
    menor = num1                # Define num1 como o menor
    # Agora compara num2 e num3 para determinar meio e maior
    if num2 <= num3:
        meio = num2
        maior = num3
    else:
        meio = num3
        maior = num2

# Caso 2: num2 é o menor (quando num2 <= num1 e num2 <= num3)
elif num2 <= num1 and num2 <= num3:
    menor = num2                # Define num2 como o menor
    # Compara num1 e num3 para ordenar os restantes
    if num1 <= num3:
        meio = num1
        maior = num3
    else:
        meio = num3
        maior = num1

# Caso 3: num3 é o menor (quando nenhum dos anteriores é verdadeiro)
else:
    menor = num3                # Define num3 como o menor
    # Compara num1 e num2 para definir meio e maior
    if num1 <= num2:
        meio = num1
        maior = num2
    else:
        meio = num2
        maior = num1

# Exibe os três números em ordem crescente, um por linha
print("\nNúmeros ordenados em ordem crescente:")
print(f"{menor}\n{meio}\n{maior}")
```
