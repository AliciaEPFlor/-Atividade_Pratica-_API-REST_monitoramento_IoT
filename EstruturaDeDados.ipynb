Adiciona o evento no final da lista (append) ou remover (pop)

```python
minha_lista = [1, 2, 3]
minha_lista.append(4)
minha_lista.append("Hello")
print(minha_lista)
minha_lista.pop()
print(minha_lista)
```

Função que muda a reordena as strings em ordem numerica/alfabetica (sort) e inverte (reverse)

```python
minha_lista = ["e", "c", "b", "a", "d"]
minha_lista.sort()
print(minha_lista)
```

```python
minha_lista = ['a', 'b', 'c', 'd', 'e']
minha_lista.reverse()
print(minha_lista)
```

```python
minha_lista = ['1', '2', '3', '4', '5']
minha_lista.reverse()
print(minha_lista)
```
